```python
messages = []
for row in rows:
    json_str = row[2]
    data = json.loads(json_str)
    type = data['data']['type']
    if type == 'human':
        type = 'user'
    else:
        type = 'assistant'
    content = data['data']['content']
    message = {'id': row[0], 'role': type, 'content': content, 'images': None}
    messages.append(message)
print(messages)
```

```python
def insert_messages_with_id_and_images(messages):
    conn = sqlite3.connect('messages.db')
    c = conn.cursor()
    for message in messages:
        c.execute('INSERT INTO messages (id, role, content, images) VALUES (?, ?, ?, ?)', 
            (message['id'], message['role'], message['content'], message['images']))
    conn.commit()
    conn.close()

insert_messages_with_id_and_images(messages)
```

```python
import sqlite3

def save_messages_to_new_db():
    # 기존 데이터베이스에서 id가 55 이상인 행을 추출
    conn = sqlite3.connect('messages.db')
    c = conn.cursor()
    c.execute('SELECT * FROM messages WHERE id >= 55')
    rows = c.fetchall()
    conn.close()

    # 새로운 데이터베이스 파일 생성 및 테이블 생성
    new_conn = sqlite3.connect('messages_0119.db')
    new_c = new_conn.cursor()
    new_c.execute('''
        CREATE TABLE IF NOT EXISTS messages (
            id INTEGER PRIMARY KEY,
            role TEXT,
            content TEXT,
            images TEXT
        )
    ''')

    # 추출한 데이터를 새로운 데이터베이스에 삽입
    for row in rows:
        new_c.execute('INSERT INTO messages (id, role, content, images) VALUES (?, ?, ?, ?)', 
                      (row[0], row[1], row[2], row[3]))
    new_conn.commit()
    new_conn.close()

# id가 55 이상인 메시지를 새로운 데이터베이스 파일로 저장
save_messages_to_new_db()
```
